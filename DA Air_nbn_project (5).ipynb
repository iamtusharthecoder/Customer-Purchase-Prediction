# Deleting Unnecessary Columns

```python
data.drop(['last review', 'reviews per month',
       'review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license'], axis = 1, inplace = True)
```

# Renamining The Columns

```python
new_column_name = []
for i in data.columns:
    new_column_name.append(i.upper())
```

# Handling Missing Values

```python
data.isnull().sum()
```

```python
data.columns
```

```python


data.dropna(subset=[ 'NAME', 'HOST IDENTITY VERIFIED', 'HOST NAME',
       'NEIGHBOURHOOD GROUP', 'NEIGHBOURHOOD', 'LAT', 'LONG', 'COUNTRY',
       'COUNTRY CODE', 'INSTANT BOOKABLE', 'CANCELLATION POLICY',
       'CONSTRUCTION YEAR', 'PRICE', 'SERVICE FEE', 'MINIMUM NIGHTS',
       'NUMBER OF REVIEWS'], inplace = True)
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
data.dtypes
```

# Removing Dollar Sign From Price Column

```python
data['PRICE'].unique().sum()
```

```python
data['PRICE'] = data['PRICE'].astype(str)
data['PRICE'] = data['PRICE'].str.replace('$', '', regex=False)
data['PRICE'] = data['PRICE'].str.replace(',', '', regex=False)
data['PRICE'] = data['PRICE'].astype(float)
```

```python
data['PRICE'].dtype
```

```python
data.head()
```

# Visualization

```python
# Histogram

sns.histplot(data['PRICE'])
plt.show()
```

```python
Insight:

Most Airbnb listings fall in the mid-price range
Presence of outliers indicates luxury listings
```

```python
# Heatmap

sns.heatmap(data.corr(numeric_only=True), annot=True)
plt.show()
```

```python
Insight:

Price has weak/moderate correlation with most variables
No single feature strongly affects price
Some variables may slightly influence pricing

Price is influenced by multiple factors, not one
Cannot predict price based on a single variable
```

```python
data.dtypes
```

```python
# Location-wise Price

sns.barplot(x='NEIGHBOURHOOD GROUP', y='PRICE', data=data)
plt.xticks(rotation=90)
plt.show()
```

```python
Insight:

Some neighbourhoods have slightly higher prices(Queens)
Differences are not very large across categories
Location/room type impacts pricing decisions
Customers can choose based on budget
```

## Key Analytical Questions And Insights

```python
# Q1. Which area is the most expensive in the dataset?
```

```python
data.groupby('NEIGHBOURHOOD GROUP')['PRICE'].mean().sort_values(ascending=False)
```

```python
Insight:

Queens has slightly higher average listing price
Prices across all neighbourhoods are almost similar
```

```python
# Q2. Which host has the most listings?

data['HOST NAME'].value_counts().head(4)
```

```python
Insight:
    
A host named "Michael" appears to have the highest listings
```

```python
# Q3. What is the average price of listings?

data['PRICE'].mean()
```

```python
Insight:

The average Airbnb listing price is around ₹625
Most listings are centered around this value
```

```python
# Q4. Which room type is most expensive?

data.groupby('ROOM TYPE')['PRICE'].mean().sort_values(ascending=False)
```

```python
Insight:
Hotel rooms have the highest average price among all room types
Shared rooms are also relatively expensive compared to entire homes and private rooms
Entire homes and private rooms have almost similar pricing
```

```python
# Q5. Top 10 expensive listings

data.sort_values(by='PRICE', ascending=False).head(10)
```

```python
Insight:

The highest priced listings are capped around 1200
Top expensive listings are distributed across multiple neighbourhoods
Both private rooms and entire homes appear in high-price category

Most listings belong to:

Brooklyn
Manhattan
Queens
```

## Conclusion and Key Insights

```python

- The Airbnb dataset analysis shows that most listings fall within a mid-price range, with an average price of around 625
- Price variation across neighbourhoods is minimal, indicating that location alone does not significantly impact pricing
- Hotel rooms are slightly more expensive compared to other room types, while private rooms and entire homes have similar pricing
- The top-priced listings are capped around a similar value, indicating possible dataset limitations or pricing constraints
- Overall, pricing is influenced by multiple factors such as room type, location, and property features rather than a single 
  variable
```
